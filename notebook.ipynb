```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
docs = [
    "I love luxeries cars, such as Mercedes BenZ",
    "Luxeries cars are very comfortable",
    "Luxeries cars don't only appear beautiful, but also it should be good in mechanical wise"
]
```

```python
embeddings = model.encode(docs)
```

```python
embeddings.shape
```

```python
query = "Mercedes"
query_embeddings = model.encode(query)
query_embeddings.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sim = cosine_similarity(embeddings , [query_embeddings])
```

```python
import numpy as np
```

```python
best_match = np.argmax(sim)
```

```python
docs[best_match]
```
